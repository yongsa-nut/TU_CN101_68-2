**Q20**: Calculate the Average GPA of a Classroom

The program works as follows:
- Ask the user how many students got grade A, B, C, D, F
- Calculate total GPA by multiplying the number of students for each grade and summing them up
- Find the total number of students
- Calculate average GPA from total GPA divided by total number of students
- If average GPA is greater than 2, display ```Good```

**Input**:
- 5 integers

**Output**:
- A real number for average GPA with two decimal places

**Note**:
- Must use **NAME CONSTANT**.
- Name your variables appropriately!

Example 1:
```
>Enter the number of A: 5
>Enter the number of B: 10
>Enter the number of C: 8
>Enter the number of D: 4
>Enter the number of F: 3
The average GPA of the classroom is: 2.33
Good
```
**Explanation**:
- The total GPA is `5*4 + 10*3 + 8*2 + 4*1 + 3*0 = 70`
- The total number of students is `5 + 10 + 8 + 4 + 3 = 30`
- The average is `70/30 = 2.33`
- 2.33 is greater than 2 so display ```good```
