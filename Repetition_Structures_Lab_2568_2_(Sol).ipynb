**Q19**: Calculate sum from user

Write a program to calculate the sum from numbers that the user enters. The program will keep asking the user to enter numbers until the user enters -999999. When finished, print the total sum.

**Input**:
- integer numbers from users

**Output**:
- The sum of all number (not include -999999)

Example 1:
```
>Enter a number: 5
>Enter a number: 3
>Enter a number: 10
>Enter a number: -999999
The total is 18
```

Example 2:
```
>Enter a number: 20  
>Enter a number: -30
>Enter a number: -999999
The total is -10.
```

````python
total = 0
num = int(input(">Enter a number:"))
while num != -999999:
  total += num
  num  = int(input(">Enter a number:"))

print(f'The total is {total}.')
````

**Q20**: Calculate salary with validation

Write a program that receives the number of hours and hourly rate, then displays the calculated salary.

For this program, you must validate that the number of hours and rate are not negative.

Hint: Check one at a time. Use two separate while loops.

**Input**:
- Real number, the number of hour.
- Real number, rate per hour.

**Output**:
- Real number, your salary at two decimal places.



Example 1:
```
>Enter hours: 20
>Enter rate: 10
Your salary is 200.00
```

Example 2:
```
>Enter hours: -10
Invalid hours. Please reenter.
>Enter hours: -20
Invalid hours. Please reenter.
>Enter hours: 30
>Enter rate: 20
Your salary is 600.00
```

Example 3:
```
>Enter hours: -25
Invalid hours. Please reenter.
>Enter hours: 40
>Enter rate: -10
Invalid rate. Please reenter.
>Enter rate: 15
Your salary is 600.00
```

````python
hours = int(input(">Enter hours: "))
while hours < 0 :
  print("Invalid hours. Please reenter.")
  hours = int(input(">Enter hours: "))

rate = float(input(">Enter rate: "))
while rate < 0 :
  print("Invalid rate. Please reenter.")
  rate = float(input(">Enter rate: "))

salary = hours*rate
print(f"Your salary is {salary:.2f}")
````
